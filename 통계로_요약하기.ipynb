## **분위수 (quartile)**

분위수는 데이터를 순서대로 늘어 놓았을 때 이를 균등한 간격으로 나누는 기준점
예를 들어 이분위수는 전체 데이터를 두 구간으로 나누기 때문에 중앙값에 해당
많이 사용하는 사분위수는 순서대로 정렬된 데이터를 네 구간으로 나눔
따라서 사분위수는 3개가 나오고 각각 25%, 50%, 75%에 해당

25%에 해당하는 값을 제 1사분위수

중앙값을 제 2사분위수

75%에 해당하는 값을 제 3사분위수


# **quantile() 메서드**

```python
# 하위 25% 구간 0.25

ns_book7['대출건수'].quantile(0.25)
```

```python
# 여러 개의 분위수 지정

ns_book7['대출건수'].quantile([0.25,0.5,0.75])

# 각 분위수 해당하는 값을 담은 시리즈 객체 반환
```

```python
# 1,2,3,4,5가 있을 때 90% 위치에 있는 값은 얼마일가
# 시리즈 객체를 정의한 후 분위수를 구하기

pd.Series([1,2,3,4,5]).quantile(0.9)

# 4.6 출력
```

quantile() 메서드는 **interpolation매개변수**에서 중간 값을 계산하는 방법 결정
위 코드는 interpolation 매개변수를 따로 지정하지 않아서 기본값으로 계산되었음
예를 들어 4 와 5 사이에서 90%에 해당하는 분위수를 찾을 때 4의 분위수 0.75와 5의 분위수 1.0을 사용한 간단한 비례식을 세울 수 있음

이렇게 두 지점 사이에 놓인 특정 위치의 값을 구하는 방법을 보간 interpolation이라고 부름

A:a = B:b
b = (axB)/A = ((0.9-0.75)x(5-4))/(1.0-0.75) = 0.6

```python
4 + (0.9-0.75) * (5-4) / (1.0-0.75)
```

```python
pd.Series([1,2,3,4,5]).quantile(0.9, interpolation = 'midpoint')
```

interpolation 매개변수는 'linear'외에도 여러 가지 다른 보간 방식을 제공

'midpoint'는 분위수에 상관없이 무조건 두 수 사이의 중앙값을 사용

## **백분위 구하기**

앞서 분위수를 직접 지정해서 구하였는데 이번에는 반대로
1,2,3,4,5의 4의 백분위는 0.75
예로 남산 도서관 대출 데이터에서 대출건수 10이 위치한 백분위를 어떻게 찾을 수 있을까
판다스에서 제공되는 메서드는 없어 간단한 트릭으로

대출건수열의 값이 10보다 작은지 비교하여 불리언배열로 만들기

```python
borrow_10_flag = ns_book7['대출건수'] < 10
```

```python
borrow_10_flag.mean()
```

데이터의 양이 많다면 전체 데이터를 순서대로 놓았을 때 10이 위치한 백분위와 거의 비슷하게 됨
따라서, 다음처럼 mean() 메서드를 호출하여 평균을 구하면 10보다 작은 값이 차지하는 비율을 간단히 얻을 수 있음

그럼 10에 대한 백분위는 0.65 정도가 됨
quantile()메서드에 백분위를 넣어 직접 확인해보면 기대했던 10이 출력

```python
ns_book7['대출건수'].quantile(0.65)
```

## **분산 (variance)**

분산은 평균으로부터 데이터가 얼마나 퍼져있는지 나타내는 통계량
데이터가 가운데 모여있다면 분산이 작고 넓게 퍼져있다면 분산이 큼


```python
ns_book7['대출건수'].var()
```

분산은 데이터가 평균에서 얼마나 퍼져있는지 나타내는 값이기 때문에 이 결과만 보고 값이 작은지 큰지 판단하기 어려움
분산은 제곱을 했기 때문에 데이터가 평균에서 멀어질수록 값이 급격히 커짐
또 단위가 대출건수의 제곱이 되어 해석하기 어려움
이를 보완하기 위해 분산에 제곱근을 취한 표준편차를 사용

## **표준편차 (standard deviation)구하기**

표준편차는 분산에 제곱근을 한 것으로 수식기호는 s
분산은 s^2

```python
ns_book7['대출건수'].std()
```

도서대출 건수는 평균 11회가 조금 넘음
어떤 도서의 대출 건수를 확인했을 때 평균보다 19만큼 더 많거나 적을 수 있다는 의미
대출건수는 음수가 될 수 없기에 평균 11, 표준편차가 19면 너무 큼
아마도 평균보다 훨씬 큰 대출건수가 있기 때문에 표준편차가 큰 것으로 예상

```python
# 수식으로 코드 구현

import numpy as np

diff = ns_book7['대출건수'] - ns_book7['대출건수'].mean()
np.sqrt(np.sum(diff**2) / len(ns_book7)-1)
```

## **데이터프레임에서 기술통계 구하기**

평균, 중앙값, 최솟값, 최댓값, 분위수, 분산, 표준편차를 구할 수 있는 다양한 판다스 메서드를 알아보았음
이런 메서드들은 데이터프레임에서도 사용 가능
수치형 열만 연산할 수 있기 때문에 해당 열에만 적용되도록 numeric_only매개변수를 True로 지정
ns_book7 df에서 mean()메서드를 호출하면 수치형인 네 개의 열에 대한 평균값이 계산
만약 numeric_only 매개변수를 지정하지 않으면 모든 데이터 타입 열에서 계산을 수행하기 때문에 시간이 매우 오래 걸리고 경고 발생

```python
ns_book7.mean(numeric_only = True)

# '번호'열은 모두 고유한 값으로 이루어져 있기 때문에 최빈값을 찾는 것이 의미가 없음
# loc메서드로 '도서명'열부터 마지막 열까지에 대해 최빈값을 찾으면 다음과 같음
```

```python
ns_book7.loc[:, '도서명':].mode()

# 도서명열에서 가장 많이 등장하는 값은 승정원일기
# 출판사열에서 가장 많이 등장하는 값은 문학동네
# 승정원일기 =! 문학동네 출판사 도서
# mode() 메서드 결과 출력물 사이엔 서로 연관 없음 주의
```
